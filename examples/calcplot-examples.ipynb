# CalcPlot Examples in Jupyter Notebook

This notebook demonstrates how to use CalcPlot functions in Jupyter environment with `show`, `explore`, and `compare`.

```typescript
import { defineIVP, simulate, show, explore, compare, view, slider } from 'npm:calcplot@0.0.4';
```

## Example 1: Basic Simulation with show()

Simple harmonic oscillator visualization using `show()`.

```typescript
// Define harmonic oscillator
const oscillator = defineIVP({
  state: { x: 1, v: 0 },
  params: { omega: 1 },
  derivatives: {
    x: (s) => s.v,
    v: (s, p) => -(p.omega**2) * s.x
  }
});

// Simulate and visualize
const timeline = simulate(oscillator, { timeRange: [0, 10] });
await show(timeline, view().plot((s) => s.x).axis('Time', 'Position').defaults());
```

## Example 2: Interactive Exploration with explore()

Damped oscillator with interactive parameter controls.

```typescript
// Define damped oscillator
const dampedOscillator = defineIVP({
  state: { x: 1, v: 0 },
  params: { omega: 1, damping: 0.1 },
  derivatives: {
    x: (s) => s.v,
    v: (s, p) => -(p.omega**2) * s.x - p.damping * s.v
  }
});

// Interactive exploration
await explore(dampedOscillator, {
  params: {
    omega: slider(0.1, 5, 1, 'Frequency'),
    damping: slider(0, 1, 0.1, 'Damping')
  },
  view: view().plot((s) => s.x).axis('Time', 'Position').defaults()
});
```

## Example 3: Comparison with compare()

Compare different damping regimes side by side.

```typescript
// Define oscillator model
const oscillatorModel = defineIVP({
  state: { x: 1, v: 0 },
  params: { omega: 1, damping: 0 },
  derivatives: {
    x: (s) => s.v,
    v: (s, p) => -(p.omega**2) * s.x - p.damping * s.v
  }
});

// Simulate different damping scenarios first
const undampedTimeline = simulate(oscillatorModel, { params: { damping: 0 }, timeRange: [0, 10] });
const lightDampingTimeline = simulate(oscillatorModel, { params: { damping: 0.1 }, timeRange: [0, 10] });
const heavyDampingTimeline = simulate(oscillatorModel, { params: { damping: 0.5 }, timeRange: [0, 10] });
const criticalDampingTimeline = simulate(oscillatorModel, { params: { damping: 2.0 }, timeRange: [0, 10] });

// Compare the timelines with proper selectors
await compare({
  'Undamped': undampedTimeline,
  'Light Damping': lightDampingTimeline,
  'Heavy Damping': heavyDampingTimeline,
  'Critical Damping': criticalDampingTimeline
}, view()
  .plot(s => s.Undamped_x, { label: 'Undamped' })
  .plot(s => s.Light_Damping_x, { label: 'Light Damping' })
  .plot(s => s.Heavy_Damping_x, { label: 'Heavy Damping' })
  .plot(s => s.Critical_Damping_x, { label: 'Critical Damping' })
  .axis({ 
    xLabel: 'Time (s)', 
    yLabel: 'Position'
  })
  .title('Damping Regimes Comparison')
);
```

## Example 4: Multiple Views with show()

Display multiple aspects of the same simulation.

```typescript
// Define pendulum model
const pendulum = defineIVP({
  state: { theta: Math.PI/4, omega: 0 },
  params: { g: 9.81, L: 1 },
  derivatives: {
    theta: (s) => s.omega,
    omega: (s, p) => -(p.g/p.L) * Math.sin(s.theta)
  }
});

// Simulate
const pendulumTimeline = simulate(pendulum, { timeRange: [0, 5] });

// Show angle over time
await show(pendulumTimeline, view().plot((s) => s.theta).axis('Time', 'Angle (rad)').defaults());
```

```typescript
// Show phase space
await show(pendulumTimeline, view().plot((s) => ({ x: s.theta, y: s.omega })).axis('Angle', 'Angular Velocity').defaults());
```

## Example 5: Complex System with explore()

Interactive predator-prey model.

```typescript
// Define Lotka-Volterra model
const predatorPrey = defineIVP({
  state: { prey: 10, predator: 5 },
  params: { alpha: 1.1, beta: 0.4, delta: 0.1, gamma: 0.4 },
  derivatives: {
    prey: (s, p) => p.alpha * s.prey - p.beta * s.prey * s.predator,
    predator: (s, p) => p.delta * s.prey * s.predator - p.gamma * s.predator
  }
});

// Interactive exploration
await explore(predatorPrey, {
  params: {
    alpha: slider(0.5, 2.0, 1.1, 'Prey growth rate'),
    beta: slider(0.1, 1.0, 0.4, 'Predation rate'),
    delta: slider(0.01, 0.5, 0.1, 'Predator efficiency'),
    gamma: slider(0.1, 1.0, 0.4, 'Predator death rate')
  },
  view: view()
    .plot((s) => s.prey, { label: 'Prey', color: 'blue' })
    .plot((s) => s.predator, { label: 'Predator', color: 'red' })
    .axis('Time', 'Population')
    .defaults()
});
```

## Example 6: Parameter Study with compare()

Compare different initial conditions.

```typescript
// Define exponential growth model
const exponentialGrowth = defineIVP({
  state: { x: 1 },
  params: { k: 0.5 },
  derivatives: {
    x: (s, p) => p.k * s.x
  }
});

// Simulate different initial conditions
const x05Timeline = simulate({ ...exponentialGrowth, state: { x: 0.5 } }, { timeRange: [0, 5] });
const x10Timeline = simulate({ ...exponentialGrowth, state: { x: 1.0 } }, { timeRange: [0, 5] });
const x20Timeline = simulate({ ...exponentialGrowth, state: { x: 2.0 } }, { timeRange: [0, 5] });
const x50Timeline = simulate({ ...exponentialGrowth, state: { x: 5.0 } }, { timeRange: [0, 5] });

// Compare with proper selectors
await compare({
  'x0_0_5': x05Timeline,
  'x0_1_0': x10Timeline,
  'x0_2_0': x20Timeline,
  'x0_5_0': x50Timeline
}, view()
  .plot(s => s.x0_0_5_x, { label: 'x₀ = 0.5' })
  .plot(s => s.x0_1_0_x, { label: 'x₀ = 1.0' })
  .plot(s => s.x0_2_0_x, { label: 'x₀ = 2.0' })
  .plot(s => s.x0_5_0_x, { label: 'x₀ = 5.0' })
  .axis('Time', 'Population')
  .title('Initial Conditions Comparison')
);
```

## Summary

This notebook demonstrated three main visualization functions in CalcPlot:

- **`show()`**: Simple visualization of a single simulation
- **`explore()`**: Interactive exploration with parameter sliders
- **`compare()`**: Side-by-side comparison of multiple scenarios

Each function provides different ways to analyze and understand dynamical systems.
